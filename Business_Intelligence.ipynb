```
# Mengonversi kolom Date menjadi datetime
final_table['Date'] = pd.to_datetime(final_table['Date'])

# Mengonversi kolom Time menjadi datetime
final_table['Time'] = pd.to_datetime(final_table['Time'], format='%H:%M:%S').dt.time

# Menambahkan kolom untuk tahun, bulan, hari, dan jam dari kolom Date
final_table['Year'] = final_table['Date'].dt.year
final_table['Month'] = final_table['Date'].dt.month
final_table['Day'] = final_table['Date'].dt.day
final_table['Weekday'] = final_table['Date'].dt.weekday

# Jika ingin mengambil jam dari kolom Time
final_table['Hour'] = pd.to_datetime(final_table['Time'].astype(str), format='%H:%M:%S').dt.hour

# Menampilkan hasil
print(final_table)
```

```
monthly_data = final_table.groupby(
    [final_table['Date'].dt.month.rename('Month'),
     final_table['Date'].dt.year.rename('Year')]
).size().reset_index(name='Transactions')

# Menggabungkan kolom Month dan Year menjadi kolom baru 'Month-Year'
monthly_data['Month-Year'] = monthly_data['Month'].astype(str) + '-' + monthly_data['Year'].astype(str)

# Pastikan 'Month-Year' terurut
monthly_data['Month-Year'] = pd.Categorical(
    monthly_data['Month-Year'],
    ordered=True,
    categories=sorted(monthly_data['Month-Year'].unique())
)

# Membuat grafik batang (bar chart)
plt.figure(figsize=(12, 6))
sns.barplot(data=monthly_data, x='Month-Year', y='Transactions', color='skyblue')

# Menambahkan judul dan label
plt.title('Transaksi per Bulan dan Tahun')
plt.xlabel('Bulan-Tahun')
plt.ylabel('Jumlah Transaksi')

# Menyesuaikan label sumbu X agar tidak bertumpuk
plt.xticks(rotation=45, ha='right')

# Menampilkan grafik
plt.grid(True)
plt.show()
```

```
import matplotlib.pyplot as plt
import seaborn as sns

# Data transaksi per hari dalam bulan tertentu (misal bulan Januari)
daily_data = final_table[final_table['Month'] == 1].groupby('Day').size().reset_index(name='Transactions')

# Membuat grafik batang (bar chart)
plt.figure(figsize=(10, 6))
bar_plot = sns.barplot(data=daily_data, x='Day', y='Transactions', color='lightcoral')

# Menambahkan angka di atas batang
for bar in bar_plot.patches:
    plt.text(
        bar.get_x() + bar.get_width() / 2,  # Posisi x (di tengah batang)
        bar.get_height() + 0.5,            # Posisi y (sedikit di atas batang)
        f'{int(bar.get_height())}',        # Teks angka di atas batang
        ha='center',                       # Horizontal alignment
        fontsize=10,                       # Ukuran font
        color='black'                      # Warna teks
    )

# Menambahkan judul dan label
plt.title('Transaksi per Hari pada Bulan Januari', fontsize=14)
plt.xlabel('Hari', fontsize=12)
plt.ylabel('Jumlah Transaksi', fontsize=12)
plt.grid(axis='y', linestyle='--', linewidth=0.7)  # Grid hanya di sumbu y

# Menampilkan grafik
plt.show()
```

```
# Grafik per jam (jam sibuk)
hourly_data = final_table.groupby('Hour').size().reset_index(name='Transactions')

# Membuat grafik
plt.figure(figsize=(10, 6))
sns.lineplot(data=hourly_data, x='Hour', y='Transactions', marker='o')
plt.title('Jam Sibuk (Transaksi per Jam)')
plt.xlabel('Jam')
plt.ylabel('Jumlah Transaksi')
plt.grid(True)
plt.show()
```

```
# Grafik berdasarkan hari dalam minggu
weekday_data = final_table.groupby('Weekday').size().reset_index(name='Transactions')

# Mengganti angka dengan nama hari
weekday_data['Weekday'] = weekday_data['Weekday'].map({
    0: 'Monday', 1: 'Tuesday', 2: 'Wednesday', 3: 'Thursday',
    4: 'Friday', 5: 'Saturday', 6: 'Sunday'
})

# Membuat grafik
plt.figure(figsize=(10, 6))
sns.barplot(data=weekday_data, x='Weekday', y='Transactions')
plt.title('Jumlah Transaksi per Hari dalam Seminggu')
plt.xlabel('Hari')
plt.ylabel('Jumlah Transaksi')
plt.grid(True)
plt.show()
```

```
# Menghitung jumlah transaksi/penjualan untuk setiap lokasi store
store_data = final_table['StoreLocation'].value_counts()

# Membuat pie chart
plt.figure(figsize=(10, 8))
explode = [0.05] * len(store_data)

plt.pie(
    x=store_data,
    labels=store_data.index,
    autopct='%1.1f%%',
    startangle=140,
    explode=explode,
    colors=plt.cm.Paired.colors,
    shadow=True,
    wedgeprops={'linewidth': 1.5, 'edgecolor': 'white'},
    textprops={'fontsize': 12, 'color': 'darkblue'}
)

# Menambahkan judul
plt.title('Distribusi Penjualan per Store', fontsize=18, color='darkblue', weight='bold')

# Menampilkan grafik
plt.show()
```

```
plt.figure(figsize=(8, 6))
sns.boxplot(y='Price', data=final_table, color='skyblue', fliersize=5)

# Menambahkan judul dan label
plt.title('Boxplot Harga per Unit', fontsize=16)
plt.ylabel('Harga per Unit', fontsize=12)

# Menampilkan grafik
plt.show()

# Menampilkan 5-number summary dalam teks
five_number_summary = {
    'Minimum': final_table['Price'].min(),
    'Q1 (25%)': final_table['Price'].quantile(0.25),
    'Median (50%)': final_table['Price'].median(),
    'Q3 (75%)': final_table['Price'].quantile(0.75),
    'Maximum': final_table['Price'].max()
}

for key, value in five_number_summary.items():
    print(f"{key}: {value:.2f}")
```

```
category_counts = final_table['ProductCategory'].value_counts().reset_index()
category_counts.columns = ['ProductCategory', 'Transaction']

# Menampilkan kategori yang paling populer
most_popular_category = category_counts.iloc[0]
print(f"Kategori paling populer: {most_popular_category['ProductCategory']} dengan {most_popular_category['Transaction']} transaksi")

# Membuat bar chart untuk visualisasi
plt.figure(figsize=(10, 6))
sns.barplot(data=category_counts, x='ProductCategory', y='Transaction', palette='viridis')

# Menambahkan judul dan label
plt.title('Distribusi Kategori Produk dalam Transaksi', fontsize=16)
plt.xlabel('Kategori Produk', fontsize=12)
plt.ylabel('Jumlah Transaksi', fontsize=12)

# Menampilkan nilai pada setiap bar
for index, row in category_counts.iterrows():
    plt.text(index, row['Transaction'], row['Transaction'], color='black', ha='center', fontsize=10)

# Menampilkan plot
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```
# Ambil kolom target
values = final_table['TotalAmount'].values

# Normalisasi data
scaler = MinMaxScaler(feature_range=(0, 1))
values = scaler.fit_transform(values.reshape(-1, 1))

# Tentukan ukuran data train-test split
percent_train = 0.8
size_data_train = int(len(values) * percent_train)

train = values[:size_data_train]
test = values[size_data_train:]

# Tentukan panjang sequence
sequence_length = 7  # 7 hari

# Fungsi untuk membuat data sequence
def create_sequences(data, sequence_length):
    x, y = [], []
    for i in range(len(data) - sequence_length):
        x.append(data[i:i + sequence_length])
        y.append(data[i + sequence_length])
    return np.array(x), np.array(y)

# Buat data sequence untuk training dan testing
x_train, y_train = create_sequences(train, sequence_length)
x_test, y_test = create_sequences(test, sequence_length)
```
